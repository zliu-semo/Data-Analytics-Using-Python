```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Download stock data for TSLA, NVDA, GOOG and AAPL
tsla_data = yf.download('TSLA', start='2024-04-01', end='2024-06-20')
nvda_data = yf.download('NVDA', start='2024-04-01', end='2024-06-20')
goog_data = yf.download('GOOG', start='2024-04-01', end='2024-06-20')
aapl_data = yf.download('AAPL', start='2024-04-01', end='2024-06-20')

# Merge the closing prices and volume into a single DataFrame
stock_data = pd.concat([tsla_data, nvda_data, goog_data, aapl_data], axis=1)
stock_data = stock_data.swaplevel(axis=1)

# Display the first few rows of the DataFrame
print(stock_data.head())
print(stock_data.index[:5])

# Create a figure and an array of subplots
fig, axes = plt.subplots(3, 2, figsize=(10, 8))
# each of the scatter plot is placed on one of the subplots
stock_data.plot(x=('TSLA', 'Close'), y=('NVDA', 'Close'), ax = axes[0, 0], title='TSLA vs NVDA Close Prices', kind = 'scatter')
stock_data.plot(x=('TSLA', 'Close'), y=('GOOG', 'Close'), ax = axes[0, 1], title='TSLA vs GOOG Close Prices', kind = 'scatter', color='orange')
stock_data.plot(x=('TSLA', 'Close'), y=('AAPL', 'Close'), ax = axes[1, 0], title='TSLA vs AAPL Close Prices', kind = 'scatter', color='green')
stock_data.plot(x=('NVDA', 'Close'), y=('GOOG', 'Close'), ax = axes[1, 1], title='NVDA vs GOOG Close Prices', kind = 'scatter', color='red')
stock_data.plot(x=('NVDA', 'Close'), y=('AAPL', 'Close'), ax = axes[2, 0], title='NVDA vs AAPL Close Prices', kind = 'scatter', color='purple')

# Add grid lines for better readability
axes[0, 0].grid(True)
axes[0, 1].grid(True)
axes[1, 0].grid(True)
axes[1, 1].grid(True)
axes[2, 0].grid(True)

# Adjust layout to prevent overlapping titles and labels
plt.tight_layout()

# Display the plot
plt.show()
```
