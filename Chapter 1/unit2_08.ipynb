### Regular Expressions

```python
"""
https://developers.google.com/edu/python/regular-expressions
"""
import re
str = 'an example word:cat!!'
match = re.search(r'word:\w\w\w', str)
#match = re.search(r'word:\w.+', str) #+: 1 or more occurrences of the pattern to the left -- 1 or more letter here
# If-statement after search() tests if it succeeded
if match:
  print('found', match.group()) ## 'found word:cat'
else:
  print('did not find')
```

```python
line = """Lions are faster than puppy 
        bull dogs"""
"""
.: single character except newline \n
*: 0 or more occurences of the pattern to the left
?: 0 or 1 occurence of the pattern to the left
(): group extraction
"""

# The re.match() function attempts to match the pattern at the start of the string (not throughout)
# (.*) matches any number of characters before the word "are". This is captured as Group 1
# ((.*?\s)*) matches 0 or more occurrences of:
# (.*?\s): A non-greedy match of any characters followed by a space. 
# The question mark (?) after the asterisk makes the quantifier non-greedy. 
# This means that it will match the fewest characters possible while still allowing the rest of the regex to match.
# The outer ((.*?\s)*) captures the entire group of words after "are". This forms Group 2.
# re.M | re.I are flags:
#     re.M (multiline): Allows ^ and $ to match the start and end of each line.
#     re.I (ignore case): Makes the match case-insensitive. 

matchObj = re.match( r'(.*) are ((.*?\s)*)', line, re.M|re.I)#multi line and ignore case
#matchObj = re.match( r'(.*) are ((.*\s)*)', line, re.M|re.I)
#matchObj = re.search( r'(.*)\sare\s(.*?)\s(.*)', line)

if matchObj:
   print("matchObj.group() : ", matchObj.group())
   print("matchObj.group(1) : ", matchObj.group(1))
   print("matchObj.group(2) : ", matchObj.group(2))
   print("matchOjb.group(3) : ", matchObj.group(3))
else:
   print("No match!!")
```

### Matching Versus Searching
Python offers two different primitive operations based on regular expressions: match checks for a match only at the beginning of the string, while search checks for a match anywhere in the string

```python
line = "Lions are faster than dogs";

matchObj = re.match( r'dogs', line, re.M|re.I)
if matchObj:
   print("match --> matchObj.group() : ", matchObj.group())
else:
   print("No match!!")

searchObj = re.search( r'dogs', line, re.M|re.I)
if searchObj:
   print("search --> searchObj.group() : ", searchObj.group())
else:
   print("Nothing found!!")
```

### Search and Replace
One of the most important re methods that use regular expressions is sub.

re.sub(pattern, repl, string, max=0)

This method replaces all occurrences of the RE pattern in string with repl, substituting all occurrences unless max provided. This method returns modified string.

```python
phone = "2009-123-567 # This is Phone Number"

# Delete Python-style comments
num = re.sub(r'#.*', "", phone)
print("Phone Num : ", num)

# Remove anything other than digits
num = re.sub(r'\D', "", phone)   #\D matches any character that is NOT a digit 
print("Phone Num : ", num)
```
